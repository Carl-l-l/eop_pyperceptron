### Matricer

```python
# Transponering af en matrix
x = np.array([[1], [2], [3]]) # Søjlevektor, inputs
w = np.array([[0.5], [0.1], [0.2]]) # Søjlevektor, vægte

z = np.dot(x.T, w) # Prikprodukt af transponerede x og w (z = vægtsum)

print("x:", x)
print('-'*20)
print("w:", w)
print('-'*20)
print("z (vægtsum): ", z)
```
